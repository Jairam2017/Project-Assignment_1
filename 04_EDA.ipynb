```python
import pandas as pd
df = pd.read_csv ("/content/drive/MyDrive/Colab Notebooks/earthquakes_cleaned.csv")
```

```python
# Top 10 Magnitude Earthquakes

top10 = df.sort_values(
    by="mag",
    ascending=False
)

print(top10[
    ["time","place","mag"]
].head(10))
```

```python
# Earthquakes by Year

print(df["year"].value_counts())
```

```python
# Earthquakes by Magnitute Category


print(df["mag_Category"].value_counts())
```

```python
# Earthquakes by Depth Category

print(
    df["depth_category"].value_counts()
)
```

```python
# Top 20 Countries

print(df["country"].value_counts().head(20))

```

```python
# Create Visualizations

# Magnitude Distribution

import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(
    df["mag"],
    bins=20,
    kde=True
    )

plt.show()
```

```python
# HEATMAP

corr = df[
    ["mag",
     "depth_km",
     "sig",
     "felt",
     "cdi",
     "mmi"
     ]].corr()

sns.heatmap(
         corr,
         annot=True,
         cmap="coolwarm"

     )
plt.show
```

```python
# Magnitude vs Depth

sns.scatterplot(
    data=df,
    x="depth_km",
    y="mag"
)

plt.show()
```
