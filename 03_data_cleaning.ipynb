```python
import pandas as pd

df=pd.read_csv("/content/drive/MyDrive/Colab Notebooks/earthquakes_raw.csv")
```

```python
# DATA CLEANING - FILLING MISSING VALUES

df["alert"] = df["alert"].fillna("None")

df["felt"] = df["felt"].fillna(0)

df["cdi"] = df["cdi"].fillna(0)

df["mmi"] = df["mmi"].fillna(0)

df["nst"] = df["nst"].fillna(df["nst"].median())

df["dmin"] = df["dmin"].fillna(df["dmin"].median())

df["rms"] = df["rms"].fillna(df["rms"].median())

df["gap"] = df["gap"].fillna(df["gap"].median())
```

```python
print(df.isnull().sum())
```

```python
# FEATURE ENGINEERING
# Convert to Datetime

df["time"] = pd.to_datetime(df["time"])

df["updated"] = pd.to_datetime(df["updated"])
```

```python
# Create year, month & day Column

df["year"] = df["time"].dt.year

df["month"] = df["time"].dt.month

df["month_name"] = df["time"].dt.month_name()

df["day"] = df["time"].dt.day

df["day_name"] = df["time"].dt.day_name()

df["hour"] = df["time"].dt.hour
```

```python
# Extract Country

import re

def extract_country(place):

  match = re.search(r',\s*(.*)$', str(place))

  if match:
    return match.group(1)

    return "Unknown"
```

```python
df["country"] = df["place"].apply(extract_country)
```

```python
# Create Magnitude Category

def magnitude_category(mag):

    if mag < 4:
        return "Minor"

    elif mag < 5:
        return "Light"

    elif mag < 6:
        return "Moderate"

    elif mag < 7:
        return "Strong"

    else:
        return "Major"
```

```python
df["mag_Category"] = df["mag"].apply(magnitude_category)
```

```python
# Create Depth Category

def depth_category(depth):

    if depth < 70:
        return "Shallow"

    elif depth < 300:
        return "Intermediate"

    else:
        return "Deep"
```

```python
df["depth_category"] = df["depth_km"].apply(depth_category)
```
